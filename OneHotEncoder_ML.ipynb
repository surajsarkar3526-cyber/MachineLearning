###What is Encoding in Machine Learning

Encoding in ML is the process of converting categorical data into numerical format so that algorithms can process it.

###OneHotEcoder

OneHotEncoder in sklearn is used to convert categorical data into a binary (0/1) matrix — one column for each category.

Used on: Nominal (non-ordinal) categorical features (e.g., color: red, green, blue). Purpose: Makes data usable for ML models that can't handle text labels directly

```python
import pandas as pd
```

```python
df = pd.read_csv("binary_classification_sample.csv")
df
```

###How to Apply Oridinal Encoding?

```python
df.dropna(inplace=True)
```

```python
df['Department'].value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
Encoder = OneHotEncoder(drop='first', sparse_output=False)
df['Department'] = Encoder.fit_transform(df[['Department']]).astype(int)
df
```

```python
cols = ['Department']
cols
```

```python
encoder = OneHotEncoder(drop='first', sparse_output=False)
encoder
```

```python
encoder_col = encoder.fit_transform(df[cols]).astype(int)
```

```python
encoder_df = pd.DataFrame(encoder_col, columns=encoder.get_feature_names_out(cols))
encoder_df
```

```python
encoder_df.index = df.index
```

```python
final_df = pd.concat([df,encoder_df],axis=1)
```
