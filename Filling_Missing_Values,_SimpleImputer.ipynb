##Filling Missing Values, SimpleImputer

##Load Dataset

```python
import pandas as pd
df = pd.read_csv("binary_classification_sample.csv")
df
```

##Filling Missing Values

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
```

```python
# fill numerical column values
impt1 = SimpleImputer(strategy='mean')
df[['Salary','LocationScore']] = impt1.fit_transform(df[['Salary','LocationScore']])
```

```python
df.isnull().sum()
```

```python
impt2 = SimpleImputer(strategy='most_frequent')

df[['Education']] = impt2.fit_transform(df[['Education']])
```
